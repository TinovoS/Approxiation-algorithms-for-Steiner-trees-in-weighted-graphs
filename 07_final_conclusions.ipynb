# Konačni zaključak — poređenje nad svim datasetovima

Testirali smo 7 algoritama na 3 dataseta različitih veličina:
- **Small** (small01–small10): 5–30 čvorova, 3–8 terminala — sa brute force verifikacijom
- **B** (b01–b18): 50–100 čvorova, 9–50 terminala
- **C** (c01–c20): 500 čvorova, do 100 terminala (testirano sa 30 i 50 terminala)

## Preporuke po scenariju

- **Mali grafovi (do 30 čvorova):** Koristiti **Brute Force** za egzaktno rešenje. Svi algoritmi završavaju ispod 1s, razlike su minimalne.
- **Srednji grafovi (50–100 čvorova):** **Zelikovsky** nudi najbolji kompromis — konzistentno blizu optimuma uz prihvatljivo vreme (~1–4s). Za maksimalan kvalitet koristiti **Zel+SPH**.
- **Veliki grafovi (500+ čvorova), brzina prioritet:** **Mehlhorn** — jedini algoritam koji ostaje ispod 1s. Kvalitet je solidan (±5% od MST heuristike).
- **Veliki grafovi, kvalitet prioritet:** **SPH** ili **Zel+SPH** — do 4% bolji od MST heuristike, ali uz vreme od 2–4 minuta po instanci.
- **Praktičan kompromis za velike grafove:** **Zelikovsky sa MST bazom** — stabilan kvalitet bez dramatičnog usporavanja.

## Ključni uvidi

1. **Teorijski aproksimacioni faktor ne predviđa praktični kvalitet.** Zelikovsky (11/6 ≈ 1.83) bi trebalo da bude bolji od SPH (faktor 2), ali u praksi SPH često daje bolje rezultate jer greedy pristup dobro eksploatiše strukturu realnih grafova.
2. **Mehlhorn je nesrazmerno brz** zahvaljujući Voronoi particiji — O(m log n) vs O(t² · n log n) za MST heuristiku. Na C datasetu je 1000–5000× brži od ostalih.
3. **Izbor baznog algoritma za Zelikovsky je bitan.** Zel+SPH daje najbolji kvalitet, ali Zel+Mehlhorn je brži bez velikog gubitka. Zel+MST je najstabilniji kompromis.
4. **Sa porastom broja terminala, razlike u kvalitetu rastu** ali i razlike u vremenu eksplodiraju — na 50 terminala i 500 čvorova, SPH traje 100–200× duže od Mehlhorn-a.
