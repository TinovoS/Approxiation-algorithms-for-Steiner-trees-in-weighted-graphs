## Zaključci

Na C datasetu sa 500 čvorova i do 30 terminala, razlike između algoritama postaju izraženije:

- **SPH** daje **najbolji prosečni kvalitet** (odnos **0.9589**) — u proseku ~4% bolji od MST heuristike. Posebno se ističe na instancama sa malo terminala (C07: -11%, C20: -11%), ali je **izuzetno spor** — do 169s po instanci, što ga čini nepraktičnim za veće setove.
- **Zelikovsky** je veoma konzistentan (odnos **0.9935**) — nikad lošiji od MST heuristike, ali retko značajno bolji. Vreme je umereno (14–132s).
- **Mehlhorn** ostaje najbrži (0.003–0.7s), ali na većim grafovima ponekad daje **lošije rezultate** od MST heuristike (C10: +6%, C18: +6%, C19: +6%).
- **Zel+SPH** kombinuje prednosti oba — prosečan odnos **0.9579**, ali uz najduže vreme (do 243s).
- **Zel+Mehlhorn** (0.9997) praktično ne poboljšava MST heuristiku, a ponekad pogoršava (C10: +6%, C18: +6%).

Ključni uvid: na grafovima sa 500 čvorova, SPH i Zel+SPH dominiraju po kvalitetu, ali sa enormnim troškom vremena. Mehlhorn je jedini algoritam koji ostaje pod 1s na svim instancama. Zelikovsky sa MST bazom nudi najbolji kompromis brzine i kvaliteta na ovom setu.
