## Zaključci

Na C datasetu sa 500 čvorova i do 50 terminala, povećanje broja terminala u odnosu na verziju sa 30 terminala otkriva skalabilnost algoritama:

- **SPH** i dalje daje najbolji kvalitet, ali vreme izvršavanja drastično raste sa brojem terminala — očekuju se vremena i preko 200s na instancama sa 50 terminala, što ga čini nepraktičnim bez paralelizacije.
- **Mehlhorn** ostaje dominantan po brzini i jedini algoritam primenljiv u realnom vremenu na grafovima ove veličine.
- **Zelikovsky** sa porastom terminala postaje sve sporiji (O(t³·n)), ali zadržava stabilan kvalitet blizu MST heuristike.
- **Zel+SPH** daje vrhunski kvalitet, ali kombinuje sporost oba algoritma — praktično neupotrebljiv bez vremenskog ograničenja.
- **Zel+Mehlhorn** profitira od brzog Mehlhorn starta, ali kvalitet zavisi od toga koliko se Mehlhorn-ova Voronoi struktura poklapa sa optimalnim stablom.

Poređenje sa verzijom od 30 terminala pokazuje da sa porastom broja terminala raste i prednost SPH-a po kvalitetu, ali uz neproporcionalan rast vremena. Za praktičnu primenu na grafovima od 500+ čvorova sa 50+ terminala, Mehlhorn je jedini realan izbor, osim ako kvalitet nije apsolutni prioritet.
